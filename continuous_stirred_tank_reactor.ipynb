# Overview
This notebook models what the expected quality of gas mixing will be inside a volume. 

The modeled scenario is an *oxgen test*, where oxygen gas is flowed into the volume at a constant rate. After some calculated amount of time, the concentration of oxygen measured at the end of the exhaust tube connected to the volume will read 100% concentration. The rise in concentration is expected to be exponential with time assuming even mixing.

The model used is a **Continuous Stirred Tank Reactor** and uses a differential mass balance equation that assumed perfect mixing inside the volume. In reality, we expect that gas will not mix evenly, and that we will measure concentration at the exhaust to rise faster due to it flowing more directly to the exit in the volume.
